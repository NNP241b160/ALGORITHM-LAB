```python
# Exercise 2
def func2(a, b):
  result = 1
  while b > 0:
    result = result * a
    b = b - 1
  return result
```

```python
# Exercise 4
def func4(a, b):
    b.sort() # Time complexity for this function is O(Nlog N) where N = len(b)
    intersect = 0
    for x in a:
        if binary_search(b, x) >= 0:
            intersect = intersect + 1
    return intersect

def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

```python
# Exercise 5
def func5(n):
    guess = 1
    while guess * guess <= n:
        if guess * guess == n:
            return guess
        guess = guess + 1
    return None
```

```python
# Sample function call: func5(16)
sec = timeit.timeit("func5(16)", globals=globals(), number=1)
ns = sec*1e9
print(f"Execution time: {ns:.0f} ns")
func5(16)
#THIS FUNCTION IS SHOWING UNDERROOT OF A GIVEN NUMBER .
```

```python
# Exercise 6
def func8(A):
    func8_helper(A, 0, len(A) -1)

def func8_helper(A, i, j):
    #Input: An array A and two indices i and j.
    n = j - 1 + 1
    if n ==2:
        if A[i] > A[j]:
            # Swap A[i] and A[j]
            A[i], A[j] = A[j], A[i]
        elif n > 2:
            m = n // 3
            func8_helper(A, i, j - m)
            func8_helper(A, i + m, j)
            func8_helper(A, i, j - m)
        return A
```

```python
#Sample function call: func8([3,1,2,7,8,6])
sec = timeit.timeit("func8([3,1,2,7,8,6])", globals=globals(), number=1)
ns = sec*1e9
print(f"Execution time: {ns:.0f} ns")
func8([3,1,2,7,8,6])
print(func8([3,1,2,7,8,6]))
```

```python
# Exercise 7
def func12(arr, k):
    arr_len = len(arr)
    k = k % arr_len
    # Initialize a new array of the same length as 'arr'
    custom_arr = [0] * arr_len

    for i in range(arr_len):
        mystery_index = (i + k) % arr_len
        custom_arr[mystery_index] = arr[i]
    return custom_arr

```

```python
#Sample function call: func12([1, 2, 3, 4, 5], 2)
sec = timeit.timeit("func12([1, 2, 3, 4, 5], 2)", globals=globals(), number=1)
ns = sec*1e9
print(f"Execution time: {ns:.0f} ns")
func12([1, 2, 3, 4, 5], 2)
```

Programs on 1-d arrays like - sum of elements of array, searching an element inarray, finding minimum and maximum element in array, count the number ofeven and odd numbers in array. For all such programs, also find the timeComplexity, compare if there are multiple methods

```python
# write a function named as sum_arr to find sum of elements in given array
def sum_arr(arr):
    sum=0
    for n in arr:
        sum=n+sum
    return sum
```

```python
#call the function
#arr = [12, 3, 4, 15,45,78,450]
arr=[12,3,4,15,45,78,450]
#print(sum_arr)
print(sum_arr(arr))

sec = timeit.timeit("sum_arr(arr)", globals=globals(), number=1)
ns = sec*1e9
print(f"Execution time: {ns:.0f} ns")
```

```python
#write a function named as max_min to find minimum (or maximum) element in an array
def max_min(arr,con):
  if con =="max":
    m=max(arr)
    return m
  else:
    n=min(arr)
    return n
```

```python
arr=[12, 1234, 45, 67, 1]
print(max_min(arr,"max"))

sec = timeit.timeit("max_min(arr,'max')", globals=globals(), number=1)
ns = sec*1e9
print(f"Execution time: {ns:.0f} ns")

```

```python
# write a function named as Counting_EvenOdd to count number of even and odd elements in an array
def Counting_EvenOdd(arr):
    even_count=0
    odd_count=0
    for num in arr:
        if num%2 == 0:
            even_count += 1
        else:
            odd_count += 1
    return even_count, odd_count
```

```python

arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]
print(Counting_EvenOdd(arr))

sec = timeit.timeit("Counting_EvenOdd(arr)", globals=globals(), number=1)
ns = sec*1e9
print(f"Execution time: {ns:.0f} ns")
```

```python
# write a function named as linear_Search to perform a linear search in 1D-array
def linear_Search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1
```

```python
# call the function linear_Search for array = [1, 3, 5, 7, 9]
array = [1, 3, 5, 7, 9]
search_element = 3
result = linear_Search(array, search_element)
if result != -1:
    print(f"Element {search_element} found at index {result}.")
else:
    print(f"Element {search_element} not found in the array.")

sec = timeit.timeit("linear_Search(arr, search_element)", globals=globals(), number=1)
ns = sec*1e9
print(f"Execution time: {ns:.0f} ns")
```
